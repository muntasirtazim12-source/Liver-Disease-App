```python
features = [col for col in train.columns if col != "Result"]

label_counts = (
    train.groupby(features, dropna=False)["Result"]
    .transform("nunique")
)

clean_train = (
    train.loc[label_counts == 1]
    .drop_duplicates()
    .reset_index(drop=True)
)

print("Rows removed because labels conflict:", (label_counts > 1).sum())
print("Cleaned shape:", clean_train.shape)
print("Remaining exact duplicates:", clean_train.duplicated().sum())
print("\nCleaned class counts:")
print(clean_train["Result"].value_counts().sort_index())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix
)

X = clean_train.drop(columns=["Result"])
y = clean_train["Result"].map({1: 1, 2: 0})  # 1 = disease, 0 = no disease

X_train, X_val, y_train, y_val = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

numeric_cols = X_train.select_dtypes(include="number").columns.tolist()
categorical_cols = X_train.select_dtypes(exclude="number").columns.tolist()

preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_cols),
    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])

model = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)
pred = model.predict(X_val)
prob = model.predict_proba(X_val)[:, 1]

dummy = DummyClassifier(strategy="most_frequent")
dummy.fit(X_train, y_train)

print("Majority baseline accuracy:", round(dummy.score(X_val, y_val), 4))
print("Logistic Regression")
print("Accuracy :", round(accuracy_score(y_val, pred), 4))
print("Precision:", round(precision_score(y_val, pred), 4))
print("Recall   :", round(recall_score(y_val, pred), 4))
print("F1-score :", round(f1_score(y_val, pred), 4))
print("ROC-AUC  :", round(roc_auc_score(y_val, prob), 4))
print("Confusion matrix (actual rows; predicted columns; order 0, 1):")
print(confusion_matrix(y_val, pred, labels=[0, 1]))
```

```python
tree_preprocessor = ColumnTransformer([
    ("numeric", SimpleImputer(strategy="median"), numeric_cols),
    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])
```

```python
from sklearn.base import clone
from sklearn.ensemble import (
    RandomForestClassifier, ExtraTreesClassifier, GradientBoostingClassifier
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix
)
from sklearn.pipeline import Pipeline
import pandas as pd

scaled_preprocessor = ColumnTransformer([
    ("numeric", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]), numeric_cols),
    ("categorical", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]), categorical_cols)
])

model_specs = [
    ("Logistic Regression", LogisticRegression(max_iter=1000), scaled_preprocessor),
    ("Decision Tree", DecisionTreeClassifier(random_state=42), tree_preprocessor),
    ("Random Forest", RandomForestClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), tree_preprocessor),
    ("Extra Trees", ExtraTreesClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), tree_preprocessor),
    ("Gradient Boosting", GradientBoostingClassifier(random_state=42), tree_preprocessor),
    ("XGBoost", XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        eval_metric="logloss", random_state=42, n_jobs=-1
    ), tree_preprocessor),
    ("KNN", KNeighborsClassifier(n_neighbors=5), scaled_preprocessor),
    ("SVM (RBF)", SVC(kernel="rbf", C=1.0), scaled_preprocessor)
]

results = []
fitted_models = {}
validation_predictions = {}

for name, classifier, prep in model_specs:
    print("Training:", name)
    candidate = Pipeline([
        ("preprocessor", clone(prep)),
        ("classifier", classifier)
    ])
    candidate.fit(X_train, y_train)

    prediction = candidate.predict(X_val)
    score = (
        candidate.predict_proba(X_val)[:, 1]
        if hasattr(candidate, "predict_proba")
        else candidate.decision_function(X_val)
    )

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_val, prediction),
        "Precision": precision_score(y_val, prediction),
        "Recall": recall_score(y_val, prediction),
        "F1": f1_score(y_val, prediction),
        "ROC-AUC": roc_auc_score(y_val, score)
    })
    fitted_models[name] = candidate
    validation_predictions[name] = prediction

display(
    pd.DataFrame(results)
    .sort_values(["Accuracy", "ROC-AUC"], ascending=False)
    .round(4)
    .reset_index(drop=True)
)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay, classification_report
import matplotlib.pyplot as plt

top_models = ["Random Forest", "Extra Trees"]

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

for name, ax in zip(top_models, axes):
    ConfusionMatrixDisplay.from_predictions(
        y_val,
        validation_predictions[name],
        labels=[0, 1],
        display_labels=["No disease", "Disease"],
        cmap="Blues",
        colorbar=False,
        ax=ax
    )
    ax.set_title(name)

plt.tight_layout()
plt.show()

for name in top_models:
    print(f"\n{name} — class-wise report")
    print(classification_report(
        y_val,
        validation_predictions[name],
        labels=[0, 1],
        target_names=["No disease", "Disease"],
        zero_division=0
    ))
```

```python
selected_features = [
    "Alkphos Alkaline Phosphotase",
    "Sgpt Alamine Aminotransferase",
    "Sgot Aspartate Aminotransferase",
    "Total Bilirubin",
    "ALB Albumin",
    "Total Protiens",
    "A/G Ratio Albumin and Globulin Ratio",
    "Direct Bilirubin"
]

models_to_compare = {
    "Random Forest": RandomForestClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ),
    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ),
    "XGBoost": XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        eval_metric="logloss", random_state=42, n_jobs=-1
    )
}

full_results = {row["Model"]: row for row in results}
comparison = []
selected_models = {}

for name, classifier in models_to_compare.items():
    full = full_results[name]
    comparison.append({
        "Model": name,
        "Features": "All 10",
        **{metric: full[metric] for metric in
           ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC"]}
    })

    candidate = Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", classifier)
    ])
    candidate.fit(X_train[selected_features], y_train)

    prediction = candidate.predict(X_val[selected_features])
    probability = candidate.predict_proba(X_val[selected_features])[:, 1]
    selected_models[name] = candidate

    comparison.append({
        "Model": name,
        "Features": "Top 8 labs",
        "Accuracy": accuracy_score(y_val, prediction),
        "Precision": precision_score(y_val, prediction),
        "Recall": recall_score(y_val, prediction),
        "F1": f1_score(y_val, prediction),
        "ROC-AUC": roc_auc_score(y_val, probability)
    })

display(pd.DataFrame(comparison).round(4))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scoring = {
    "Accuracy": "accuracy",
    "Precision": "precision",
    "Recall": "recall",
    "F1": "f1",
    "ROC-AUC": "roc_auc"
}

cv_specs = [
    ("Random Forest / all 10", RandomForestClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), "all"),
    ("Random Forest / top 8", RandomForestClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), "top8"),
    ("Extra Trees / all 10", ExtraTreesClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), "all"),
    ("Extra Trees / top 8", ExtraTreesClassifier(
        n_estimators=300, random_state=42, n_jobs=-1
    ), "top8"),
    ("XGBoost / all 10", XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        eval_metric="logloss", random_state=42, n_jobs=-1
    ), "all"),
    ("XGBoost / top 8", XGBClassifier(
        n_estimators=300, max_depth=4, learning_rate=0.05,
        subsample=0.9, colsample_bytree=0.9,
        eval_metric="logloss", random_state=42, n_jobs=-1
    ), "top8")
]

cv_results = []

for name, classifier, feature_set in cv_specs:
    print("Cross-validating:", name)

    if feature_set == "top8":
        X_cv = X_train[selected_features]
        prep = SimpleImputer(strategy="median")
    else:
        X_cv = X_train
        prep = clone(tree_preprocessor)

    pipeline = Pipeline([
        ("preprocessor", prep),
        ("classifier", classifier)
    ])

    scores = cross_validate(
        pipeline, X_cv, y_train,
        cv=cv, scoring=scoring, n_jobs=1
    )

    row = {"Model": name}
    for metric in scoring:
        row[metric] = (
            f"{scores['test_' + metric].mean():.4f} "
            f"+/- {scores['test_' + metric].std():.4f}"
        )
    cv_results.append(row)

display(pd.DataFrame(cv_results))
```

```python
import numpy as np
import pandas as pd

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_selection import SelectFromModel
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import StratifiedKFold, GridSearchCV
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix
)

# Target: 1 = liver disease, 0 = no liver disease
X_eval = clean_train.drop(columns="Result").copy()
y_eval = clean_train["Result"].map({1: 1, 2: 0})

if y_eval.isna().any():
    raise ValueError("Result-এ 1 বা 2 ছাড়া অন্য মান আছে—আগে সেটা পরীক্ষা করতে হবে।")

numeric_cols = X_eval.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = [c for c in X_eval.columns if c not in numeric_cols]

transformers = []
if numeric_cols:
    transformers.append((
        "numeric",
        SimpleImputer(strategy="median"),
        numeric_cols
    ))
if categorical_cols:
    transformers.append((
        "categorical",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", OneHotEncoder(handle_unknown="ignore"))
        ]),
        categorical_cols
    ))

preprocessor = ColumnTransformer(transformers=transformers)

pipeline = Pipeline([
    ("preprocess", preprocessor),
    ("select", SelectFromModel(
        RandomForestClassifier(
            n_estimators=100,
            random_state=42,
            n_jobs=1
        ),
        threshold="median"
    )),
    ("model", RandomForestClassifier(
        random_state=42,
        n_jobs=1
    ))
])

param_grid = {
    "select__threshold": ["mean", "median"],
    "model__min_samples_leaf": [1, 2]
}

outer_cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
inner_cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=123)

oof_pred = np.empty(len(y_eval), dtype=int)
oof_prob = np.empty(len(y_eval), dtype=float)
fold_rows = []

for fold, (train_idx, valid_idx) in enumerate(outer_cv.split(X_eval, y_eval), start=1):
    search = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grid,
        scoring="f1",
        cv=inner_cv,
        n_jobs=-1,
        refit=True
    )

    search.fit(X_eval.iloc[train_idx], y_eval.iloc[train_idx])

    pred = search.predict(X_eval.iloc[valid_idx])
    prob = search.predict_proba(X_eval.iloc[valid_idx])[:, 1]

    oof_pred[valid_idx] = pred
    oof_prob[valid_idx] = prob

    actual = y_eval.iloc[valid_idx]
    fold_rows.append({
        "fold": fold,
        "accuracy": accuracy_score(actual, pred),
        "precision": precision_score(actual, pred, zero_division=0),
        "recall": recall_score(actual, pred, zero_division=0),
        "f1": f1_score(actual, pred, zero_division=0),
        "roc_auc": roc_auc_score(actual, prob)
    })

fold_df = pd.DataFrame(fold_rows)
print("Per-fold scores:")
print(fold_df.round(4).to_string(index=False))

print("\nMean ± standard deviation across outer folds:")
for metric in ["accuracy", "precision", "recall", "f1", "roc_auc"]:
    print(f"{metric}: {fold_df[metric].mean():.4f} ± {fold_df[metric].std():.4f}")

print("\nPooled out-of-fold scores:")
print(f"Accuracy : {accuracy_score(y_eval, oof_pred):.4f}")
print(f"Precision: {precision_score(y_eval, oof_pred, zero_division=0):.4f}")
print(f"Recall   : {recall_score(y_eval, oof_pred, zero_division=0):.4f}")
print(f"F1-score : {f1_score(y_eval, oof_pred, zero_division=0):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_eval, oof_prob):.4f}")
print("Confusion matrix [[TN, FP], [FN, TP]]:")
print(confusion_matrix(y_eval, oof_pred, labels=[0, 1]))
```
